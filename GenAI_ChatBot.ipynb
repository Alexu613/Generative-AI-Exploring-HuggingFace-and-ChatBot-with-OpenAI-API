### Premier pas avec OpenAI :

    Voici ce que fait le code ci-dessous : 
Ce script Python permet de se connecter à l’API d’OpenAI de façon sécurisée via une saisie masquée de la clé API avec `getpass`, puis d’envoyer une requête personnalisée à ChatGPT (GPT-4). Il utilise deux entrées : un `system_prompt` pour définir le rôle ou le comportement de l’IA, et un `user_prompt` pour poser la question. Trois paramètres clés contrôlent le style de réponse : `temperature` (niveau de créativité), `max_tokens` (longueur de la réponse) et `top_p` (choix des mots les plus probables). Le script affiche ensuite la réponse générée, tout en gérant les erreurs de connexion de manière simple.

```python
#Importations des librairies:
import pandas as pd
import numpy as np
import matplotlib as plt
import warnings
warnings.filterwarnings('ignore')

#Importation de la librairies OpenAI:
import openai
#Importation de la librairie Password (Hidden):
import getpass

API_KEY = "sk-proj..."

'''
Paramètres:
-----------
temperature: Gérer la créativité	Réponses sûres	Réponses originales
max_tokens:	 Limiter la longueur de sortie	Réponses courtes	Réponses longues
top_p:       Contrôler la variété des mots	Mots sûrs	Mots diversifiés
'''

MAX_TOKENS = 300 #Limiter le nombre de mots/tokens (Taille de la réponse donnée par l'IA.
TEMPERATURE = 0 #Plus c'est haut, plus l'IA devient créative dans réponse.
TOP_P = 0.9 #Gérer le choix intelligent des mots ici, l'IA prend les 10% des idées/mots les plus sûrs (pioches les mots les plus probables).

#Insérer le contexte et le comportement de l'agent IA:
system_prompt = input('Insert system_prompt :')

#Insérer votre demande/prompt à l'IA:
user_prompt = input('Insert your prompt for IA:')

#OPENAI_API_KEY do not include " ":
def CONNECTION_TO_OPENAI_API(system, prompt):
    try :
        OPENAI_API_KEY = getpass.getpass('Enter your password :')

            # Créer un client avec ta clé API
        client = openai.OpenAI(api_key=OPENAI_API_KEY)  # Mets ici ta clé API
        
        # Envoi d’un message à ChatGPT
        response = client.chat.completions.create(
            model="gpt-4",  # ou "gpt-3.5-turbo"
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ],

            max_tokens = MAX_TOKENS,
            top_p = TOP_P,
            temperature=TEMPERATURE
            
    )
        
        # Affichage de la réponse
        print(response.choices[0].message.content)
        
    except Exception as error:
            print('Error Password: Please try a new password')
    else: 
            print('Connection Activated !')

CONNECTION_TO_OPENAI_API(system_prompt, user_prompt)
```

## Agents Systems
    Développement d'un système d'agents autonomes: 

Définir deux agents IA distincts (agent1 et agent2), chacun avec :

    - un system_prompt → qui donne à l’agent une "personnalité" ou un rôle,

    - un user_prompt (pour agent1) ou une entrée issue d’un autre agent (pour agent2).

L’agent 1 traite une demande utilisateur, produit une réponse (response_a1).

L’agent 2 reçoit la réponse de l’agent 1 comme entrée, et réagit à celle-ci avec son propre contexte.

Tu contrôles leur comportement via :

    - temperature (créativité),

    - top_p (diversité des mots),

    - max_tokens (longueur de réponse).

```python
'''
Paramètres:
-----------
temperature: Gérer la créativité	0Réponses sûres	2Réponses originales/inventés
max_tokens:	 Limiter la longueur de sortie	Réponses courtes	Réponses longues
top_p:       Contrôler la variété des mots	0Mots sûrs	1Mots diversifiés
'''

MAX_TOKENS = 300 #Limiter le nombre de mots/tokens (Taille de la réponse donnée par l'IA.
TEMPERATURE = 0 #Plus c'est haut, plus l'IA devient créative dans réponse.
TOP_P = 0.9 #Gérer le choix intelligent des mots ici, l'IA prend les 10% des idées/mots les plus sûrs (pioches les mots les plus probables).

#Insérer le contexte et le comportement de l'agent 1:
system_prompt_agent1 = input('Insert system_prompt_agent1 :')
#Insérer votre demande/prompt à l'agent 1:
user_prompt_agent1 = input('Insert your prompt for Agent 1:')
#Insérer le contexte et le comportement de l'agent 1:
system_prompt_agent2 = input('Insert system_prompt_agent2 :')


# Créer un client avec ta clé API
client = openai.OpenAI(api_key=OPENAI_API_KEY)  # Mets ici ta clé API
        
# Création de l'agent intelligent 1:
response_agent1 = client.chat.completions.create(
    model="gpt-4",  # ou "gpt-3.5-turbo"
    messages=[
            {"role": "system", "content": system_prompt_agent1},
            {"role": "user", "content": user_prompt_agent1}
        ],
    

    max_tokens = MAX_TOKENS,
    top_p = TOP_P,
    temperature=TEMPERATURE
            
)
print("----------------------AGENT INTELLIGENT 1----------------------")    
# Affichage de la réponse de l'AGENT 1:
print(f"\nRécupération de la réponse générée de \n -------------------------------------:")
print(f"Agent 1: {response_agent1.choices[0].message.content}")
response_a1 = response_agent1.choices[0].message.content


# Création de l'agent intelligent 2:
response_agent2 = client.chat.completions.create(
    model="gpt-4", 
    messages=[
        {'role':'system', 'content': system_prompt_agent2}, #Spécifier le contexte et donner une tâche à l'agent 2.
        {'role':'user', 'content': response_a1}
    ],
    max_tokens = MAX_TOKENS,
    top_p = TOP_P,
    temperature=TEMPERATURE
)
print("----------------------AGENT INTELLIGENT 2----------------------")
# Affichage de la réponse de l'AGENT 2:
print(f"\nRécupération de la réponse générée de \n -------------------------------------:")
print(f"Agent 2: {response_agent2.choices[0].message.content}")
```

Definition: Asking the model to do a task without giving any examples.
Example:

    "Translate this sentence into French: 'I love pizza.'"
    The model has to understand what "translate" means just from the instruction.

```python
import openai

client = openai.OpenAI(api_key=API_KEY)

system = "Tu es un agent intelligent capable de faire de la classification de sentiment"
prompt = """ Classifie les avis suivants par "Positif" ou "Négatif":
            1) Le film était superbe et j'ai vraiment adoré les acteurs.
            2) Ouai bof, il y'a mieux que ça.
            3) Je pense que le producteur aurait pu apporter plus de suspense.
            4) Oui, j'adore l'histoire. 
                """

response = client.chat.completions.create(
    model='gpt-4o-mini', 
    messages=[
        {"role":"system", "content": system},
        {"role":"user", "content": prompt}
    ], 
    max_tokens=30,
    temperature=0,
    top_p=None
)
print(response.choices[0].message.content)
```

### Adding assistant messages

L’assistant, c’est le rôle joué par l’intelligence artificielle (comme ChatGPT) dans une conversation.

🔹 Et les autres rôles ?

### Creating a conversation history

- In an AI conversation system, a conversation history is a structured record of the entire dialogue between a user and the AI assistant. Each message in the conversation has a specific role that defines who is speaking: the user, the assistant, or the system. This structure helps the AI understand the context of the conversation, follow along, and respond more accurately.

- The user role represents what the person says — questions, requests, or instructions. The assistant role is used for the AI’s responses. The system role is special: it sets the tone or behavior of the assistant at the start of the conversation. For example, a system message might tell the assistant to be formal, friendly, or explain things like a teacher. These roles are used together to guide the flow of the interaction and maintain coherence.

- By storing the conversation history, the AI can “remember” what was said earlier and give more relevant answers. For example, if the user asks a follow-up question, the assistant can look back at previous messages to understand what the user is referring to. This makes the conversation feel more natural and human-like, instead of starting from scratch every time.

```python
import os 
import openai
from time import sleep
from tqdm import tgrange, tqdm_notebook

SYSTEM_MESSAGE = "You are an expert in Math"
PROMPT_MESSAGE = "Can you explain to me what silouhette score for KMeans Clustering is ?"

messages=[
        {"role":"system", "content": SYSTEM_MESSAGE},
        {"role":"user", "content": PROMPT_MESSAGE}]

client = openai.OpenAI(api_key=API_KEY, organization=None, project=None, base_url=None, timeout=None)
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=messages,
    max_tokens=150,
    top_p=0.5,
    temperature=0.2,
    stop=["\n"]
)

print(response.choices[0].message.content)

#Store response to create a conversation history:
assistant_dict = {"role":"assistant", "content":response.choices[0].message.content}
messages.append(assistant_dict)
print(f"\nConversation history:")
print("---------LOADING---------")
print(messages)
        
```

### Creating a AI Chatbot

    🤖 What is a chatbot?

A chatbot is a computer program that can simulate a conversation with a human. It can answer questions, provide information, or help users do tasks — all through text (or voice).

    💡 Why build one with OpenAI API?

Using the OpenAI API lets you create very advanced chatbots that understand natural language, follow context, and generate smart, human-like answers — with very little code. It’s fast to build, easy to customize, and super powerful for customer service, education, content creation, and more.

#### A. First Step (simple version):

Let's create a simple AI Chatbot using the OpenAI API.

    Parameters :
- pi_key: Your secret key to access the AI (like a magic password).
- organization: Optional name of your company or group.
- project: Optional name of your current project or task.
- base_url: The internet address where you send your messages (default if not set).
- websocket_base_url: Special address for live, real-time communication.
- timeout: How long to wait for a reply before giving up.
- max_retries: How many times to try again if something fails (default: 2).
- default_headers: Extra info sent with every message (like “I speak English”).
- default_query: Default settings or options added to every request.
- http_client: The tool used to send messages online (can use a custom one).
- _strict_response_validation: If true, checks every reply strictly for correct format (default: false).

```python
import os 
import openai
from time import sleep
from tqdm import tgrange, tqdm_notebook

MESSAGES = [{"role":"system", "content": "You are a genius in math with an expertise in Machine Learning."}]
USER_PROMPT = ["Can you explain to me what silouhette score for KMeans Clustering is ?",
                   "Summarize this explanation in two bullet points"]
try:
    for question in tqdm(USER_PROMPT, desc="Generation Response From AI Agent"):
        if not question: 
            raise ValueError("No questions provided, please provide questions")
        else:
            print(f"User_question : {question}")
            print("--------------------------------------------------")
            user_question = {"role":"user", "content": question}
            #Add the question into the list:
            MESSAGES.append(user_question)
        
            #Pass MESSAGES through the LLM to get response from the AI Agent:
            client = openai.OpenAI(organization=None,
                                   project=None, 
                                   timeout=60*5, #5 minutes and break after that
                                   max_retries=2,
                                   api_key=API_KEY)
            
            #Response from the Intelligent Agent:
            response = client.chat.completions.create(
                model='gpt-4o-mini',
                messages=MESSAGES,
                max_tokens=150,
                temperature=0,
                top_p=0.9,
                stop=["\n"]
            )

            print(f"Assistant: {response.choices[0].message.content}")
            
            #Creating a conversation history:
            assistant_dict = {"role":"assistant", "content": response.choices[0].message.content}
            MESSAGES.append(assistant_dict) # Keep tracking the context and previous AI Agent's responses.
            print("***************************Conversation History***************************")
            print(MESSAGES)
            print("**************************************************************************")
            
except Exception as e:
    print("Error :", e)
```

### Speech-to-Text Transcription with Whisper

Speech-to-Text Transcription with Whisper refers to converting spoken language into written text using Whisper, an automatic speech recognition (ASR) system developed by OpenAI.
Key points:

    - Whisper is an open-source model.

    - It can transcribe audio in multiple languages.

    - It can also do translation from other languages to English.

It’s trained on a large amount of diverse audio data, making it good at understanding real-world accents, background noise, and varied speaking styles.

When we use "rb" in open(), it means:

    "r" = read → you're opening the file to read it

    "b" = binary → you're reading it in binary mode (not text)

    "rb" = read binary

This is used for files like audio, video, or images, where the content isn't plain text — it's raw data (1s and 0s).

So Python reads the file exactly as it is, without trying to turn it into readable text.

```python
#Import libraries: 
import openai
from tqdm import trange, tqdm

DESKTOP_PATH = os.path.expanduser("~/Desktop")
FILE_PATH = os.path.join(DESKTOP_PATH, "audio_whatsapp.mp3")
PROMPT = "My wife texted me the grocery list"

def Speech2TextTranscriptionWithWhisper(input_path, PROMPT):
    client = openai.OpenAI(api_key=API_KEY) #Load API_KEY
    #Open the OpenAI-Audio.mp3 file:
    audio_file = open(input_path, "rb")
    #Create transcript from the audio file:
    audio_response = client.audio.transcriptions.create(
        model="whisper-1", 
        file=audio_file,
        prompt=PROMPT)
    print(f"\n{audio_response.text} \n ----------------------")
    

    chat_response =client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role":"system", "content": "You are a helpful system capable of detecting the language"},
            {"role":"user", "content": "Identify the language used in this following text :" + audio_response.text}
        ]
    )
    return chat_response.choices[0].message.content
    
Speech2TextTranscriptionWithWhisper(FILE_PATH, PROMPT)
```

### Pretrained HuggingFace Tokenizer + OpenAI API

🧠 What Is a Pretrained Hugging Face Tokenizer?

Let’s say you want a computer to read and understand a sentence.
Before it can do that, you need to convert the text into something the computer can work with.
This is where a tokenizer comes in.
🔍 Step 1: What is a Tokenizer?

A tokenizer is a tool that splits up a sentence into small pieces called tokens.
These pieces can be:

    Whole words → "Hello world" → ["Hello", "world"]

    Or parts of words → "unbelievable" → ["un", "believ", "able"]

    Think of it like cutting a sentence into LEGO pieces, so the AI can understand how to build meaning from it.

🤖 What Does “Pretrained” Mean?

A pretrained tokenizer means it has already learned how to split up words based on a huge amount of text — like books, websites, and articles.
You don’t need to train it yourself — it’s ready to use.
🤝 What Is Hugging Face?

Hugging Face is a company that offers free, powerful tools for working with language models (like GPT, BERT, etc.).
They provide tokenizers and models that are already trained and widely used in AI applications — like chatbots, translation, text summarization, and more.
🔧 In Summary:

    A tokenizer breaks down sentences into smaller chunks (tokens).

    A pretrained tokenizer has already learned how to do this well.

    Hugging Face provides these tools, so you don’t have to build them from scratch.

    It’s the first step in preparing text for an AI model to understand and respond.

```python
#pip install transformers
from transformers import pipeline
from transformers import AutoTokenizer, AutoModelForSequenceClassification
from transformers import TextClassificationPipeline
from tqdm import tqdm, trange
import openai

#Simple model for sentiment-analysis:
def SimpleClassifier_From_HuggingFace(input_text):
    classifier=pipeline('sentiment-analysis', model='distilbert/distilbert-base-uncased-finetuned-sst-2-english')
    result=classifier(input_text)
    return result

#AutoTokenizer:
def AutoTokenizerClassifier(input_text):
    #Tokenizer:
    tokenizer=AutoTokenizer.from_pretrained("distilbert/distilbert-base-uncased-finetuned-sst-2-english")
    model=AutoModelForSequenceClassification.from_pretrained("distilbert/distilbert-base-uncased-finetuned-sst-2-english")
    #print(tokenizer, model)
    classifier=TextClassificationPipeline(model=model,
                                          tokenizer=tokenizer,
                                          truncation=True)
    result = classifier(input_text)
    return result
```

```python
### Encoder for TensorFlow or PyTorch:
print(f"{tokenizer.tokenize("This cake was so delicious, this is the best cake I ever tasted in my life")}")

print(f"{tokenizer.encode("This cake was so delicious, this is the best cake I ever tasted in my life")}")

print(f"{tokenizer.decode(tokenizer.encode("This cake was so delicious, this is the best cake I ever tasted in my life"))}")

tokens = tokenizer("This cake was so delicious, this is the best cake I ever tasted in my life",return_tensors="pt")
print(tokens)
```

```python
#Run the function:
SimpleClassifier_From_HuggingFace("This cake was so delicious, this is the best cake I ever tasted in my life")
```

```python
#Run the function:    
AutoTokenizerClassifier("This cake wasn't so delicious, this is one of the worst cake I ever tasted in my life")
```
